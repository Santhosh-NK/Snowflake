```
%%sql -r dataframe_1
CREATE OR REPLACE FILE FORMAT CSV_PAYMENTS_FORMAT
    TYPE = CSV

    -- Compression
    COMPRESSION = AUTO

    -- File structure
    RECORD_DELIMITER = '\n'
    FIELD_DELIMITER = ','
    MULTI_LINE = TRUE
    FILE_EXTENSION = 'csv'

    -- Header handling
    PARSE_HEADER = FALSE
    SKIP_HEADER = 1
    SKIP_BLANK_LINES = TRUE

    -- Date and time handling
    DATE_FORMAT = AUTO
    TIME_FORMAT = AUTO
    TIMESTAMP_FORMAT = AUTO

    -- Binary data
    BINARY_FORMAT = HEX

    -- Quotes, spaces and escape characters
    FIELD_OPTIONALLY_ENCLOSED_BY = '"'
    TRIM_SPACE = TRUE
    ESCAPE = NONE
    ESCAPE_UNENCLOSED_FIELD = NONE

    -- NULL and empty values
    NULL_IF = ('NULL', 'null', 'N/A', '')
    EMPTY_FIELD_AS_NULL = TRUE

    -- Validation and invalid characters
    ERROR_ON_COLUMN_COUNT_MISMATCH = TRUE
    REPLACE_INVALID_CHARACTERS = FALSE

    -- Encoding
    SKIP_BYTE_ORDER_MARK = TRUE
    ENCODING = 'UTF8'

    COMMENT = 'CSV file format for payment files';
```
