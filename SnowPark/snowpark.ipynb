```
# Basic syntax
from snowflake.snowpark import Session
from snowflake.snowpark.functions import col

df = (
    session.table("CUSTOMERS") # session is important to connect to the warehouse
    .filter(col("STATUS") == "ACTIVE")
    .select("CUSTOMER_ID", "CUSTOMER_NAME")
)

df.show()
```

```
session.close() # must be important at the end 
```

```
session.use_role("DATA_ENGINEER_ROLE")
session.use_warehouse("DE_WH")
session.use_database("DEMO_DB")
session.use_schema("CURATED")
```
